### Select Rated Features

```python
rated_features = user_movies[features] 
user_ratings = user_movies["rating"] 
rated_features 
user_ratings
```

### Weight Features using Ratings

```python
weighted_features = rated_features.mul( user_ratings, axis=0
) 
weighted_features
```

### Building the User Profile 

```python
user_profile = weighted_features.sum() 
user_profile
```

### Interpretation of the User Profile

```python
user_profile.sort_values( ascending=False )
```

The top two genres for this user are Action and SciFi with scores of 14 and 10 respectively. This suggests that the user enjoys those types of movies more because they were rated higher.

### Normalise the User Profile

```python
user_profile_normalized = ( user_profile / user_profile.sum() ) 
user_profile_normalized
```

### Scoring every movie

```python
movies["recommendation_score"] = ( movie_features .mul(user_profile_normalized) .sum(axis=1) ) 
movies[["title", "recommendation_score"]]
```

### Ranking of Movies

```python
ranked_movies = movies.sort_values( "recommendation_score", ascending=False ) 
ranked_movies[["title", "recommendation_score"]]
```

### Removal of watched Movies

```python
watched_movies = ratings["movie_id"] 
recommendations = ranked_movies[ ~ranked_movies["movie_id"].isin(watched_movies) ] 
recommendations[["title", "recommendation_score"]]
```

### Top 3 Recommendations

```python
top_3 = recommendations[ ["title", "recommendation_score"] ].head(3) 
top_3
```

The top 3 recommendations, firstly are all new movies to the user. They are Space Warriors, City Detectives and Broken Hearts all of which contain some of the users preferred features such as Action, SciFi and a bit of Drama

### Building of a Reusable Function

```python
def recommend_movies(ratings, movies, features, n=3):
    user_movies = ratings.merge(movies, on="movie_id")
    rated_features = user_movies[features]
    weighted_features = rated_features.mul(user_movies["rating"], axis=0)

    user_profile = weighted_features.sum()
    user_profile = user_profile / user_profile.sum()

    results = movies.copy()
    results["recommendation_score"] = (
        results[features].mul(user_profile).sum(axis=1)
    )

    watched = ratings["movie_id"]
    results = results.loc[~results["movie_id"].isin(watched)]

    return results.sort_values(
        "recommendation_score", ascending=False
    ).head(n)


recommend_movies(ratings, movies, features)
```

### Test on a different user

### Limitations

1. A large amount of film information is ignored such as actors, directors, release year, langauge etc. all of which could influence a user's preference profile

2. The model relies too heavily on ratings as the only indicator of enjoyment so the resulting recommemdations may become too similar to previous liked items resulting in boredom

3. New users with little or no preference history may pose a challenge as the model has no basis onto which it can build a preference profile, this is similar to the cold-start problem
